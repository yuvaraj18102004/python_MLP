```python
import pandas as pd
```

```python
d1=pd.read_csv(r"C:\Users\subas\OneDrive\Desktop\sales_data_sample.csv")
d1
```

```python
#Calculate Total Sales Amount for each transaction using a new column
d1['Total Sales']=d1['Quantity'] * d1['UnitPrice']
```

```python
d1
```

```python
d1['Region'].unique()
```

```python
#Find total sales by each Region.

d1.groupby(by='Region')['Total Sales'].sum()
```

```python
#Display all transactions where Discount is greater than 5%.
d1.loc[d1['Discount']>=5 ,['Discount']]
```

```python
#Group data by Category and find the average UnitPrice for each.
d1.groupby('Category')['UnitPrice'] .mean()
```

```python
#Find the top 2 SalesPersons with the highest total sales amount.
d1.groupby(by='SalesPerson')['Total Sales'] .sum() .nlargest(2)
```

```python
#Filter records for only Electronics category where sales are above ₹50,000.
d1[(d1['Category']=='Electronics') & (d1['Total Sales'] >=50000)]
```

```python
#Sort the dataset by Date and SalesPerson in ascending order.
d1.groupby(by='Date')['SalesPerson'].sum() .sort_values()
```

```python
d1.sort_values(by=['Date', 'SalesPerson'])
```

```python
#Find the number of transactions for each CustomerType (Regular vs New).

d1.groupby('CustomerType').size()
```

```python
#Get the most frequently sold product in each region.

d1.groupby(['Region','Product']).size() .reset_index(name='Count')
```

```python
#Pivot Table: Create a pivot table showing total sales by Region and Category

pd.pivot_table(data=d1,index=['Region'],columns=['Category'],values='Total Sales',aggfunc='sum').round(2)
```
