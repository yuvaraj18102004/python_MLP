```python
for i in range (0,6):
    for j in range (0,6):
        print("*",end=" ")
    print()
```

```python
for i in range (1,6,1):
    for j in range (1,6,1):
        if(i==1 or i==5 or j==1 or j==5):
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
for i in range (1,6,1):
    for j in range (1,i+1,1):
        print("*",end=" ")
    print()
```

```python
for i in range (1,6,1):
    for j in range (1,j,1):
        print("*",end=" ")
    print()
```

```python
n=5
for i in range(n, 0, -1):
    for j in range(n - i):
        print(" ", end=" ")
    for k in range(i):
        print("*", end=" ")
    
    print()
```
